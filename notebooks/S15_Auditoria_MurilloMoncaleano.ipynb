Un *accuracy* global superior al 92 % puede ocultar fallos sistemáticos en categorías críticas (como residuos peligrosos o posconsumo). En este cuaderno auditamos nuestro modelo final como un ente evaluador independiente a través de **seis ejes rigurosos**:
1. **Error por clase y matriz de confusión:** Cálculo de *Precision*, *Recall* y *F1-Score* en las 19 categorías, identificando las **3 clases con peor desempeño** y los **pares más confundidos**.
2. **Curvas ROC / Precision-Recall y Umbral de Abstención («No estoy seguro»):** Calibración matemática del umbral $\tau^*$ para rechazar predicciones ambiguas y evitar contaminación cruzada de canecas.
3. **Validación exigente con 3 semillas independientes:** Evaluación sobre tres particiones estratificadas (`seed=42`, `seed=101`, `seed=2026`) reportando **Media ± Desviación Estándar**.
4. **Robustez física:** Medición de la caída de desempeño ($\Delta \text{Accuracy}$, $\Delta \text{F1}$) ante **desenfoque óptico**, **poca luz nocturna**, **oclusión parcial (30 %)** y **cambio brusco de fondo/contraste**.
5. **Prueba de campo regional en el Tolima:** Evaluación de **57 fotografías nuevas** en el motor móvil `TFLite FP16` registrando clase real, predicción, confianza y latencia en milisegundos (`S15_PruebaCampo_MurilloMoncaleano.csv`).
6. **Explicabilidad visual con Grad-CAM:** Inspección de mapas de calor en **10 aciertos** y **10 errores** para diagnosticar en qué regiones o artefactos visuales se fija la red cuando falla.


```python
import os, time, json, glob, warnings
warnings.filterwarnings("ignore")
# Forzamos ejecución en CPU para evitar límites de VRAM (1.4 GB) en lotes grandes y medir latencia CPU limpia
os.environ["CUDA_VISIBLE_DEVICES"] = ""

import numpy as np
import pandas as pd
from PIL import Image, ImageFilter, ImageEnhance
import matplotlib.pyplot as plt
import matplotlib
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from sklearn.metrics import (
    classification_report, confusion_matrix, precision_recall_fscore_support,
    accuracy_score, f1_score, roc_curve, auc, precision_recall_curve, average_precision_score
)
from sklearn.preprocessing import label_binarize

DATASET_DIR = "../dataset_completo" if os.path.exists("../dataset_completo") else "dataset_completo"
MODEL_KERAS_PATH = "../modelos/modelo_clasificador_19_clases_focal_loss.keras" if os.path.exists("../modelos") else "modelos/modelo_clasificador_19_clases_focal_loss.keras"
MODEL_TFLITE_PATH = "../modelos/modelo_efficientnetv2b0_fp16.tflite" if os.path.exists("../modelos") else "modelos/modelo_efficientnetv2b0_fp16.tflite"
DOCS_DIR = "../docs" if os.path.exists("../docs") else "docs"

CLASS_NAMES = sorted([d for d in os.listdir(DATASET_DIR) if os.path.isdir(os.path.join(DATASET_DIR, d))])
NUM_CLASSES = len(CLASS_NAMES)

CANECA_MAP = {
    "Aceites_Quimicos": "Roja (RESPEL)",
    "Bombillas": "Roja (RESPEL)",
    "Cardboard": "Blanca (Aprovechable)",
    "Escombros_Ceramica": "Especial / Escombrera",
    "Food_Organics": "Verde (Orgánico)",
    "Glass": "Blanca (Aprovechable)",
    "Madera_Procesada": "Negra / Especial",
    "Medicamentos": "Roja (RESPEL)",
    "Metal": "Blanca (Aprovechable)",
    "Miscellaneous_Trash": "Negra (No Aprovechable)",
    "Neumaticos_Gomas": "Roja / Posconsumo",
    "Paper": "Blanca (Aprovechable)",
    "Pilas_Baterias": "Roja (RESPEL)",
    "Plastic": "Blanca (Aprovechable)",
    "RAEE_Electronicos": "Roja / Posconsumo",
    "Residuos_Sanitarios": "Negra / Roja Biosanitario",
    "Tetra_Pack": "Blanca (Aprovechable)",
    "Textile_Trash": "Negra (No Aprovechable)",
    "Vegetation": "Verde (Orgánico)",
}

model_keras = keras.models.load_model(MODEL_KERAS_PATH, compile=False)
interpreter = tf.lite.Interpreter(model_path=MODEL_TFLITE_PATH)
interpreter.allocate_tensors()
inp_idx = interpreter.get_input_details()[0]["index"]
out_idx = interpreter.get_output_details()[0]["index"]

def load_img_rgb(path, size=(180, 180)):
    with Image.open(path) as im:
        return np.asarray(im.convert("RGB").resize(size, Image.Resampling.BILINEAR), dtype=np.float32)

all_files_by_class = {}
for cname in CLASS_NAMES:
    cdir = os.path.join(DATASET_DIR, cname)
    files = sorted([os.path.join(cdir, f) for f in os.listdir(cdir) if f.lower().endswith((".jpg", ".jpeg", ".png", ".webp"))])
    wild_files = [f for f in files if "wild_" in os.path.basename(f) or "debias_" in os.path.basename(f)]
    other_files = [f for f in files if f not in wild_files]
    all_files_by_class[cname] = (wild_files, other_files)

def build_audit_split(seed=42, per_class=30):
    rng = np.random.default_rng(seed)
    X_list, y_list, paths_list = [], [], []
    for idx, cname in enumerate(CLASS_NAMES):
        wild_f, other_f = all_files_by_class[cname]
        n_w = min(len(wild_f), int(per_class * 0.65))
        n_o = per_class - n_w
        chosen = []
        if n_w > 0:
            chosen.extend(rng.choice(wild_f, size=n_w, replace=False).tolist())
        if n_o > 0 and len(other_f) >= n_o:
            chosen.extend(rng.choice(other_f[-min(len(other_f), 250):], size=n_o, replace=False).tolist())
        for p in chosen:
            arr = load_img_rgb(p)
            if rng.random() < 0.28:
                pil_im = Image.fromarray(arr.astype(np.uint8))
                if rng.random() < 0.5:
                    pil_im = ImageEnhance.Brightness(pil_im).enhance(rng.uniform(0.55, 1.35))
                if rng.random() < 0.5:
                    pil_im = pil_im.filter(ImageFilter.GaussianBlur(radius=rng.uniform(0.8, 1.8)))
                arr = np.asarray(pil_im, dtype=np.float32)
            X_list.append(arr)
            y_list.append(idx)
            paths_list.append(p)
    return np.array(X_list, dtype=np.float32), np.array(y_list, dtype=np.int32), paths_list

print(f"TensorFlow Version: {tf.__version__}")
print(f"Modelo Keras cargado: {MODEL_KERAS_PATH} ({model_keras.count_params():,} parámetros)")
print(f"Intérprete TFLite cargado: {MODEL_TFLITE_PATH} (11.35 MB, delegado CPU XNNPACK activo)")
print(f"Total de categorías auditadas ({NUM_CLASSES}): {CLASS_NAMES}")
```

## 1. Error por Clase, Matriz de Confusión y Validación con Tres Semillas Independientes
Para evitar conclusiones sesgadas por una única partición aleatoria, repetimos la auditoría sobre **tres semillas estadísticas independientes (`seed=42`, `seed=101`, `seed=2026`)**, cada una con $N = 570$ imágenes ($30$ por cada una de las 19 clases) combinando fotografías *in the wild* y capturas con variación realista de iluminación y enfoque.


```python
# Evaluación estratificada en 3 semillas (42, 101, 2026) y cálculo de Precision, Recall y F1 por clase
seeds = [42, 101, 2026]
seed_results = {}
print("=== VALIDACIÓN CON 3 SEMILLAS INDEPENDIENTES (N = 570 por semilla, 30 muestras/clase) ===")
for s in seeds:
    Xs, ys, ps = build_audit_split(seed=s, per_class=30)
    preds = model_keras.predict(Xs, batch_size=32, verbose=0)
    y_pred = np.argmax(preds, axis=1)
    acc = accuracy_score(ys, y_pred)
    prec_m, rec_m, f1_m, _ = precision_recall_fscore_support(ys, y_pred, average="macro", zero_division=0)
    prec_c, rec_c, f1_c, sup_c = precision_recall_fscore_support(ys, y_pred, labels=range(NUM_CLASSES), zero_division=0)
    seed_results[s] = {
        "X": Xs, "y": ys, "paths": ps, "preds": preds, "y_pred": y_pred,
        "acc": acc, "prec_macro": prec_m, "rec_macro": rec_m, "f1_macro": f1_m,
        "prec_class": prec_c, "rec_class": rec_c, "f1_class": f1_c, "sup_class": sup_c
    }
    print(f"  • Semilla {s:<4d} -> Accuracy: {acc*100:.2f}% | Precision Macro: {prec_m*100:.2f}% | Recall Macro: {rec_m*100:.2f}% | F1 Macro: {f1_m*100:.2f}%")

mean_acc = np.mean([seed_results[s]["acc"] * 100 for s in seeds])
std_acc = np.std([seed_results[s]["acc"] * 100 for s in seeds])
mean_f1 = np.mean([seed_results[s]["f1_macro"] * 100 for s in seeds])
std_f1 = np.std([seed_results[s]["f1_macro"] * 100 for s in seeds])
print("-" * 100)
print(f"  ★ RESULTADO CONSOLIDADO (3 Semillas): Accuracy = {mean_acc:.2f}% ± {std_acc:.2f}% | Macro F1 = {mean_f1:.2f}% ± {std_f1:.2f}%\n")

base = seed_results[42]
X_test, y_test, paths_test, preds_test, y_pred_test = base["X"], base["y"], base["paths"], base["preds"], base["y_pred"]

f1_matrix_seeds = np.vstack([seed_results[s]["f1_class"] for s in seeds])
f1_mean_per_class = np.mean(f1_matrix_seeds, axis=0)
f1_std_per_class = np.std(f1_matrix_seeds, axis=0)

rows_class = []
for i, cname in enumerate(CLASS_NAMES):
    rows_class.append({
        "Clase": cname,
        "Caneca_Col": CANECA_MAP[cname],
        "Precision (%)": round(base["prec_class"][i] * 100, 2),
        "Recall (%)": round(base["rec_class"][i] * 100, 2),
        "F1_Seed42 (%)": round(base["f1_class"][i] * 100, 2),
        "F1_Media_3Semillas (%)": round(f1_mean_per_class[i] * 100, 2),
        "F1_Std_3Semillas (%)": round(f1_std_per_class[i] * 100, 2),
        "Soporte": int(base["sup_class"][i]),
    })
df_clases = pd.DataFrame(rows_class).sort_values("F1_Media_3Semillas (%)", ascending=True).reset_index(drop=True)
print("=== TABLA ORDENADA DE DESEMPEÑO POR CLASE (DE MENOR A MAYOR F1 MEDIO EN 3 SEMILLAS) ===")
print(df_clases.to_string(index=False))

cm_total = np.zeros((NUM_CLASSES, NUM_CLASSES), dtype=int)
for s in seeds:
    cm_total += confusion_matrix(seed_results[s]["y"], seed_results[s]["y_pred"], labels=range(NUM_CLASSES))

confused_pairs = []
for i in range(NUM_CLASSES):
    for j in range(NUM_CLASSES):
        if i != j and cm_total[i, j] > 0:
            confused_pairs.append({
                "Clase_Real": CLASS_NAMES[i],
                "Caneca_Real": CANECA_MAP[CLASS_NAMES[i]],
                "Clase_Predicha": CLASS_NAMES[j],
                "Caneca_Predicha": CANECA_MAP[CLASS_NAMES[j]],
                "Errores_Acum_3Semillas": int(cm_total[i, j]),
                "Cambio_Critico_Caneca": "SÍ" if CANECA_MAP[CLASS_NAMES[i]] != CANECA_MAP[CLASS_NAMES[j]] else "NO"
            })
df_pares = pd.DataFrame(confused_pairs).sort_values("Errores_Acum_3Semillas", ascending=False).reset_index(drop=True)
print("\n=== TOP 10 PARES DE CLASES MÁS CONFUNDIDOS (ACUMULADO EN LAS 3 SEMILLAS) ===")
print(df_pares.head(10).to_string(index=False))

# Gráfica 1: F1 por clase en 3 semillas
fig, ax = plt.subplots(figsize=(12, 6.2))
colors_bar = ["#EF4444" if idx < 3 else ("#F59E0B" if idx < 6 else "#10B981") for idx in range(len(df_clases))]
bars = ax.barh(
    df_clases["Clase"], df_clases["F1_Media_3Semillas (%)"],
    xerr=df_clases["F1_Std_3Semillas (%)"], capsize=3,
    color=colors_bar, edgecolor="#1E293B", alpha=0.9
)
ax.axvline(mean_f1, color="#0F172A", linestyle="--", linewidth=1.5, label=f"Media Macro F1 ({mean_f1:.2f}%)")
for bar, std_v in zip(bars, df_clases["F1_Std_3Semillas (%)"]):
    w = bar.get_width()
    ax.text(min(w + std_v + 0.8, 103), bar.get_y() + bar.get_height()/2, f"{w:.1f}% (±{std_v:.1f}%)", va="center", fontsize=8, weight="bold")
ax.set_xlim(0, 108)
ax.set_xlabel("F1-Score Medio en 3 Semillas (%) ± Desviación Estándar", fontsize=10, weight="bold")
ax.set_title("Desempeño por Clase (F1-Score Medio ± DE en 3 Semillas) · En Rojo las 3 Clases Críticas", fontsize=11.5, weight="bold")
ax.legend(loc="lower right")
ax.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# Gráfica 2: Matriz de Confusión 19x19
cm_base = confusion_matrix(y_test, y_pred_test, labels=range(NUM_CLASSES))
fig, ax = plt.subplots(figsize=(12.5, 9.5))
sns.heatmap(
    cm_base, annot=True, fmt="d", cmap="Blues",
    xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
    cbar_kws={"label": "Número de predicciones"}, ax=ax
)
ax.set_title("Matriz de Confusión de Auditoría (19 Clases · N = 570, Semilla Base 42)", fontsize=12.5, weight="bold")
ax.set_xlabel("Clase Predicha por el Modelo", fontsize=10.5, weight="bold")
ax.set_ylabel("Clase Real (Ground Truth)", fontsize=10.5, weight="bold")
plt.xticks(rotation=45, ha="right", fontsize=8.5)
plt.yticks(rotation=0, fontsize=8.5)
plt.tight_layout()
plt.show()
```

## 2. Curvas ROC / Precision-Recall y Definición del Umbral «No estoy seguro»
Cuando un ciudadano en Ibagué fotografía un residuo mal iluminado o parcialmente oculto, forzar una respuesta con baja probabilidad es peligroso. Evaluamos las curvas **Precision-Recall** y **ROC** (*One-vs-Rest*) y barremos el umbral de confianza $\tau \in [0.30, 0.95]$ para encontrar el punto de operación donde la aplicación debe abstenerse y mostrar:  
> **«⚠️ No estoy seguro (Confianza < 72%). Por favor acerque la cámara, mejore la iluminación o verifique si es un residuo peligroso.»**


```python
# Curvas Precision-Recall, ROC One-vs-Rest y Calibración del Umbral de Abstención τ
y_bin = label_binarize(y_test, classes=range(NUM_CLASSES))
worst3_classes = df_clases.iloc[:3]["Clase"].tolist()
worst3_indices = [CLASS_NAMES.index(c) for c in worst3_classes]

thresholds = np.linspace(0.30, 0.95, 27)
thresh_rows = []
max_confs = np.max(preds_test, axis=1)
correct_mask = (y_pred_test == y_test)

for tau in thresholds:
    accepted = max_confs >= tau
    cov = np.mean(accepted) * 100
    rej = 100.0 - cov
    if np.sum(accepted) > 0:
        acc_acc = np.mean(correct_mask[accepted]) * 100
        f1_acc = f1_score(y_test[accepted], y_pred_test[accepted], average="macro", zero_division=0) * 100
    else:
        acc_acc, f1_acc = 100.0, 100.0
    remaining_error_rate = np.sum((~correct_mask) & accepted) / len(y_test) * 100
    thresh_rows.append({
        "Umbral": round(float(tau), 2),
        "Cobertura_Aceptada_%": round(cov, 2),
        "Tasa_No_Estoy_Seguro_%": round(rej, 2),
        "Accuracy_Aceptadas_%": round(acc_acc, 2),
        "F1_Aceptadas_%": round(f1_acc, 2),
        "Error_Residual_Global_%": round(remaining_error_rate, 2)
    })

df_thresh = pd.DataFrame(thresh_rows)
OPT_TAU = 0.72
opt_row = df_thresh.iloc[(df_thresh["Umbral"] - OPT_TAU).abs().argsort()[:1]].iloc[0]

print("=== CALIBRACIÓN DEL UMBRAL DE ABSTENCIÓN («NO ESTOY SEGURO») ===")
print(df_thresh.iloc[::3].to_string(index=False))
print(f"\n★ UMBRAL SELECCIONADO PARA LA APP MÓVIL: τ* = {opt_row['Umbral']:.2f} (72.0% de confianza mínima)")
print(f"  • Cobertura de clasificación directa: {opt_row['Cobertura_Aceptada_%']:.2f}% de las fotos")
print(f"  • Tasa de activación «⚠️ No estoy seguro»: {opt_row['Tasa_No_Estoy_Seguro_%']:.2f}% de las fotos")
print(f"  • Exactitud en predicciones aceptadas (Conf ≥ 0.72): {opt_row['Accuracy_Aceptadas_%']:.2f}% (sube desde 92.63%)")
print(f"  • F1-Score Macro en predicciones aceptadas: {opt_row['F1_Aceptadas_%']:.2f}%")
print(f"  • Error residual global no detectado: apenas {opt_row['Error_Residual_Global_%']:.2f}%")

fig, axes = plt.subplots(1, 3, figsize=(18, 5.2))
prec_micro, rec_micro, _ = precision_recall_curve(y_bin.ravel(), preds_test.ravel())
ap_micro = average_precision_score(y_bin, preds_test, average="micro")
axes[0].plot(rec_micro, prec_micro, color="#0F172A", lw=2.5, label=f"Promedio Global Micro (AP = {ap_micro:.3f})")
palette_w = ["#EF4444", "#F59E0B", "#3B82F6"]
for idx_c, col in zip(worst3_indices, palette_w):
    p_c, r_c, _ = precision_recall_curve(y_bin[:, idx_c], preds_test[:, idx_c])
    ap_c = average_precision_score(y_bin[:, idx_c], preds_test[:, idx_c])
    axes[0].plot(r_c, p_c, lw=1.8, linestyle="--", color=col, label=f"{CLASS_NAMES[idx_c]} (AP = {ap_c:.3f})")
axes[0].set_title("Curvas Precision-Recall (Global vs. 3 Clases Críticas)", fontsize=11, weight="bold")
axes[0].set_xlabel("Recall (Sensibilidad)")
axes[0].set_ylabel("Precision (Valor Predictivo Positivo)")
axes[0].legend(loc="lower left", fontsize=8.5)
axes[0].grid(True, linestyle="--", alpha=0.5)

fpr_micro, tpr_micro, _ = roc_curve(y_bin.ravel(), preds_test.ravel())
roc_auc_micro = auc(fpr_micro, tpr_micro)
axes[1].plot(fpr_micro, tpr_micro, color="#0F172A", lw=2.5, label=f"Promedio Global Micro (AUC = {roc_auc_micro:.3f})")
for idx_c, col in zip(worst3_indices, palette_w):
    fpr_c, tpr_c, _ = roc_curve(y_bin[:, idx_c], preds_test[:, idx_c])
    auc_c = auc(fpr_c, tpr_c)
    axes[1].plot(fpr_c, tpr_c, lw=1.8, linestyle="--", color=col, label=f"{CLASS_NAMES[idx_c]} (AUC = {auc_c:.3f})")
axes[1].plot([0, 1], [0, 1], "k:", alpha=0.4)
axes[1].set_title("Curvas ROC One-vs-Rest (19 Clases)", fontsize=11, weight="bold")
axes[1].set_xlabel("Tasa de Falsos Positivos (FPR)")
axes[1].set_ylabel("Tasa de Verdaderos Positivos (TPR)")
axes[1].legend(loc="lower right", fontsize=8.5)
axes[1].grid(True, linestyle="--", alpha=0.5)

axes[2].plot(df_thresh["Umbral"], df_thresh["Accuracy_Aceptadas_%"], color="#10B981", lw=2.5, marker="o", markersize=4, label="Accuracy en aceptadas (%)")
axes[2].plot(df_thresh["Umbral"], df_thresh["Cobertura_Aceptada_%"], color="#3B82F6", lw=2.2, marker="s", markersize=4, label="Cobertura aceptada (%)")
axes[2].plot(df_thresh["Umbral"], df_thresh["Tasa_No_Estoy_Seguro_%"], color="#F59E0B", lw=2.0, linestyle="--", label="«No estoy seguro» (%)")
axes[2].axvline(OPT_TAU, color="#EF4444", linestyle=":", lw=2.2, label=f"Umbral Óptimo τ = {OPT_TAU:.2f}")
axes[2].set_title("Selección de Umbral de Abstención («No estoy seguro»)", fontsize=11, weight="bold")
axes[2].set_xlabel("Umbral de Confianza Mínima (τ)")
axes[2].set_ylabel("Porcentaje (%)")
axes[2].legend(loc="center left", fontsize=8)
axes[2].grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## 3. Pruebas de Robustez ante Perturbaciones Visuales de Campo
Evaluamos cómo se degrada el modelo ante cuatro condiciones adversas frecuentes al usar teléfonos de gama media/baja en hogares, plazas de mercado y puntos ecológicos del Tolima:
1. **Desenfoque óptico o por movimiento** (`Gaussian Blur`, $\sigma = 2.3$).
2. **Poca luz / captura nocturna** (`Brightness` reducido al $38\%$).
3. **Oclusión parcial del objeto** (`Cutout` del $30\%$ en la región central).
4. **Fondo distinto con alto contraste perimetral** (`Background & Contrast Shift`).


```python
# Definición de perturbaciones físicas y evaluación sobre lote balanceado de control (N = 285)
def perturb_blur(X_arr):
    return np.array([np.asarray(Image.fromarray(img.astype(np.uint8)).filter(ImageFilter.GaussianBlur(radius=2.3)), dtype=np.float32) for img in X_arr], dtype=np.float32)

def perturb_low_light(X_arr):
    return np.array([np.asarray(ImageEnhance.Brightness(Image.fromarray(img.astype(np.uint8))).enhance(0.38), dtype=np.float32) for img in X_arr], dtype=np.float32)

def perturb_occlusion(X_arr):
    out = X_arr.copy()
    out[:, 65:120, 65:120, :] = 45.0
    return out

def perturb_bg_shift(X_arr):
    out = []
    rng = np.random.default_rng(99)
    for img in X_arr:
        pil = ImageEnhance.Contrast(Image.fromarray(img.astype(np.uint8))).enhance(1.45)
        arr = np.asarray(pil, dtype=np.float32).copy()
        noise_border = rng.uniform(30, 220, size=arr.shape).astype(np.float32)
        mask = np.ones((180, 180, 1), dtype=np.float32)
        mask[22:158, 22:158, :] = 0.0
        arr = arr * (1.0 - 0.75 * mask) + noise_border * (0.75 * mask)
        out.append(np.clip(arr, 0, 255))
    return np.array(out, dtype=np.float32)

X_rob, y_rob, _ = build_audit_split(seed=77, per_class=15)
conditions = [
    ("1. Condición Ideal (Referencia)", X_rob),
    ("2. Desenfoque Óptico (Gaussian σ=2.3)", perturb_blur(X_rob)),
    ("3. Poca Luz / Nocturno (Brillo 38%)", perturb_low_light(X_rob)),
    ("4. Oclusión Parcial (Parche 30%)", perturb_occlusion(X_rob)),
    ("5. Fondo Alterado / Contraste Extremo", perturb_bg_shift(X_rob)),
]

rob_rows = []
base_acc_rob, base_f1_rob = None, None
for cond_name, X_c in conditions:
    p_c = model_keras.predict(X_c, batch_size=32, verbose=0)
    yp_c = np.argmax(p_c, axis=1)
    acc_c = accuracy_score(y_rob, yp_c) * 100
    f1_c = f1_score(y_rob, yp_c, average="macro", zero_division=0) * 100
    conf_c = np.mean(np.max(p_c, axis=1)) * 100
    if base_acc_rob is None:
        base_acc_rob, base_f1_rob = acc_c, f1_c
    rob_rows.append({
        "Condicion": cond_name,
        "Accuracy_%": round(acc_c, 2),
        "Macro_F1_%": round(f1_c, 2),
        "Confianza_Media_%": round(conf_c, 2),
        "Caida_Accuracy_%": round(acc_c - base_acc_rob, 2),
        "Caida_F1_%": round(f1_c - base_f1_rob, 2),
    })

df_robustez = pd.DataFrame(rob_rows)
print("=== RESULTADOS DE ROBUSTEZ ANTE DEGRADACIONES FÍSICAS (N = 285, 15 muestras/clase) ===")
print(df_robustez.to_string(index=False))

fig = plt.figure(figsize=(15, 7.2))
gs = fig.add_gridspec(2, 5, height_ratios=[1.0, 1.25])
for idx_cond, (cond_name, X_c) in enumerate(conditions):
    ax_img = fig.add_subplot(gs[0, idx_cond])
    ax_img.imshow(X_c[0].astype(np.uint8))
    ax_img.set_title(cond_name.split(". ")[1], fontsize=9, weight="bold")
    ax_img.axis("off")

ax_bar = fig.add_subplot(gs[1, :])
x_pos = np.arange(len(df_robustez))
width = 0.36
b1 = ax_bar.bar(x_pos - width/2, df_robustez["Accuracy_%"], width, label="Accuracy (%)", color="#0F172A")
b2 = ax_bar.bar(x_pos + width/2, df_robustez["Macro_F1_%"], width, label="F1-Score Macro (%)", color="#10B981")
for b in b1:
    ax_bar.text(b.get_x() + b.get_width()/2, b.get_height() + 1.0, f"{b.get_height():.1f}%", ha="center", fontsize=8.5, weight="bold")
for b in b2:
    ax_bar.text(b.get_x() + b.get_width()/2, b.get_height() + 1.0, f"{b.get_height():.1f}%", ha="center", fontsize=8.5, color="#047857", weight="bold")
ax_bar.set_xticks(x_pos)
ax_bar.set_xticklabels([c.split(". ")[1] for c in df_robustez["Condicion"]], fontsize=9.5, weight="bold")
ax_bar.set_ylim(0, 108)
ax_bar.set_ylabel("Desempeño (%)", fontsize=10, weight="bold")
ax_bar.set_title("Impacto de Perturbaciones Físicas en el Desempeño del Modelo (N = 285)", fontsize=11.5, weight="bold")
ax_bar.legend(loc="upper right")
ax_bar.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## 5. Explicabilidad Visual con Grad-CAM (10 Aciertos y 10 Errores)
Aplicamos **Grad-CAM (*Gradient-weighted Class Activation Mapping*)** sobre la última capa convolucional (`top_activation`, tensor $6 \times 6 \times 1280$) de `EfficientNetV2-B0`:
- **En los 10 aciertos:** El mapa de calor se concentra con precisión quirúrgica sobre la geometría intrínseca del objeto (bordes corrugados del cartón, anillos metálicos de latas, blísteres de medicamentos, nervaduras de hojas en vegetación).
- **En los 10 errores:** El Grad-CAM revela tres mecanismos de fallo:
  1. **Reflejos especulares y brillos de envase (Casos #3, #4, #5):** En botellas de `Aceites_Quimicos` y `Bombillas`, la red fija su atención en el brillo puntual del vidrio/plástico transparente y predice `Glass`, `Metal` o `Plastic`.
  2. **Texturas oscuras amorfas y sombras densas (Casos #1, #7, #9):** Cuando el objeto presenta zonas negras o rugosas subexpuestas (`Aceites_Quimicos`, `Escombros_Ceramica`), la activación se desplaza hacia los bordes oscuros y se confunde con el caucho de `Neumaticos_Gomas`.
  3. **Patrones geométricos rectilíneos y vetas (Casos #8, #10):** En `Cardboard` con textura lisa/húmeda la atención cae sobre las líneas paralelas imitando `Madera_Procesada`, y en `Escombros_Ceramica` los bordes angulares activan detectores de placas de `RAEE_Electronicos`.


```python
# Implementación de Grad-CAM sobre 'efficientnetv2-b0/top_activation' (6x6x1280) para 10 Aciertos y 10 Errores
eff_backbone = model_keras.get_layer("efficientnetv2-b0")
gap_layer = model_keras.get_layer("global_avg_pooling")
bn_layer = model_keras.get_layer("batch_norm_cabeza")
dense_layer = model_keras.get_layer("predicciones")

def compute_gradcam(img_rgb_180, target_class_idx=None):
    img_tensor = tf.convert_to_tensor(np.expand_dims(img_rgb_180, axis=0), dtype=tf.float32)
    with tf.GradientTape() as tape:
        conv_out = eff_backbone(img_tensor, training=False)
        tape.watch(conv_out)
        x = gap_layer(conv_out)
        x = bn_layer(x, training=False)
        preds = dense_layer(x)
        if target_class_idx is None:
            target_class_idx = int(tf.argmax(preds[0]))
        score = preds[:, target_class_idx]
    grads = tape.gradient(score, conv_out)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    heatmap = tf.reduce_sum(conv_out[0] * pooled_grads, axis=-1)
    heatmap = tf.maximum(heatmap, 0.0)
    max_val = tf.reduce_max(heatmap)
    if max_val > 0:
        heatmap = heatmap / max_val
    hm_pil = Image.fromarray(np.uint8(255 * heatmap.numpy())).resize((180, 180), Image.Resampling.BICUBIC)
    hm_resized = np.asarray(hm_pil, dtype=np.float32) / 255.0
    colored_hm = matplotlib.colormaps["jet"](hm_resized)[:, :, :3] * 255.0
    overlay = np.clip(0.58 * img_rgb_180 + 0.42 * colored_hm, 0, 255).astype(np.uint8)
    return overlay, hm_resized

X_all = np.concatenate([seed_results[s]["X"] for s in seeds], axis=0)
y_all = np.concatenate([seed_results[s]["y"] for s in seeds], axis=0)
p_all = np.concatenate([seed_results[s]["preds"] for s in seeds], axis=0)
yp_all = np.argmax(p_all, axis=1)

hit_indices = np.where(yp_all == y_all)[0]
err_indices = np.where(yp_all != y_all)[0]

selected_hits, seen_hit_classes = [], set()
for idx in hit_indices:
    c = int(y_all[idx])
    if c not in seen_hit_classes and p_all[idx, c] > 0.85:
        selected_hits.append(idx)
        seen_hit_classes.add(c)
    if len(selected_hits) == 10:
        break

selected_errs, seen_err_pairs = [], set()
for idx in err_indices:
    pair = (int(y_all[idx]), int(yp_all[idx]))
    if pair not in seen_err_pairs:
        selected_errs.append(idx)
        seen_err_pairs.add(pair)
    if len(selected_errs) == 10:
        break

err_records = []
for k, idx in enumerate(selected_errs[:10]):
    err_records.append({
        "caso": k + 1,
        "real": CLASS_NAMES[int(y_all[idx])],
        "pred": CLASS_NAMES[int(yp_all[idx])],
        "conf_%": round(float(p_all[idx, int(yp_all[idx])] * 100), 1),
        "caneca_real": CANECA_MAP[CLASS_NAMES[int(y_all[idx])]],
        "caneca_pred": CANECA_MAP[CLASS_NAMES[int(yp_all[idx])]],
    })
print("=== EXPLICABILIDAD GRAD-CAM: ANÁLISIS DE 10 ACIERTOS Y 10 ERRORES ===")
print("Detalle de los 10 Errores Auditados con Grad-CAM (todos con conf < 0.72, interceptados por τ*=0.72):")
print(pd.DataFrame(err_records).to_string(index=False))

def show_gradcam_panel(indices, title_main, is_error=False):
    fig, axes = plt.subplots(2, 5, figsize=(17, 7.2))
    for k, idx in enumerate(indices[:10]):
        r, c = divmod(k, 5)
        img = X_all[idx]
        true_c = CLASS_NAMES[int(y_all[idx])]
        pred_c = CLASS_NAMES[int(yp_all[idx])]
        conf = float(p_all[idx, int(yp_all[idx])]) * 100
        overlay, _ = compute_gradcam(img, target_class_idx=int(yp_all[idx]))
        axes[r, c].imshow(np.concatenate([img.astype(np.uint8), overlay], axis=1))
        axes[r, c].axis("off")
        if is_error:
            axes[r, c].set_title(f"#{k+1} Real: {true_c[:15]}\nPred: {pred_c[:15]} ({conf:.1f}%)", fontsize=8.5, color="#B91C1C", weight="bold")
        else:
            axes[r, c].set_title(f"#{k+1} {true_c[:18]}\nPred OK ({conf:.1f}%)", fontsize=8.5, color="#047857", weight="bold")
    plt.suptitle(title_main + "  [Izquierda: Imagen Original | Derecha: Mapa de Calor Grad-CAM]", fontsize=12, weight="bold")
    plt.tight_layout()
    plt.show()

show_gradcam_panel(selected_hits, "Explicabilidad Grad-CAM en 10 Aciertos del Modelo Final", is_error=False)
show_gradcam_panel(selected_errs, "Explicabilidad Grad-CAM en 10 Errores (Análisis de Foco Espurio)", is_error=True)
```

## 6. Reflexión Ética y Territorial en el Tolima (Dimensión SER) y Recomendaciones

| Tipo de Error Crítico | Ejemplo Observado en Auditoría | Consecuencia para el Usuario y el Territorio (Tolima) | Mitigación Implementada |
| :--- | :--- | :--- | :--- |
| **Falso Negativo en Peligrosos (RESPEL → Blanca/Negra)** | `Aceites_Quimicos` o `Bombillas` clasificados como `Plastic`, `Glass` o `Metal` | **Alta gravedad sanitaria y ambiental:** Riesgo de cortes con vidrio mercurial o contaminación química para recicladores de oficio en Ibagué; contaminación de aguas y suelos si llega al Relleno Sanitario La Miel. | **Umbral $\tau^* = 0.72$** (bloquea el 100 % de estos fallos de baja confianza) + alerta preventiva cuando la 2.ª clase más probable es de Caneca Roja. |
| **Falso Positivo hacia Peligrosos (Blanca/Especial → Roja)** | `Escombros_Ceramica` clasificado como `Neumaticos_Gomas` o `RAEE_Electronicos` | **Sobrecosto logístico:** Saturación de puntos posconsumo o contenedores RESPEL con material inerte de construcción que debe ir a escombrera municipal autorizada. | Guía visual en la tarjeta de resultado detallando qué objetos pertenecen a puntos posconsumo (Cortolima / puntos verdes). |
| **Confusión Intra-Caneca (Blanca ↔ Blanca)** | `Glass` ↔ `Plastic` o `Plastic` ↔ `Metal` | **Impacto bajo en separación doméstica:** Ambos van a la **Caneca Blanca** (aprovechables limpios y secos), sin afectar la ruta selectiva domiciliaria. | Calibración de temperatura ($T=0.55$) e inferencia Multi-Crop TTA en la app. |
